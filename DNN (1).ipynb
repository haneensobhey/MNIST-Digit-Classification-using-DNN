```python
(Xtrain, Ytrain), (Xtest, Ytest) = mnist.load_data()
from keras.utils import to_categorical
Ytest=to_categorical(Ytest)
Ytrain=to_categorical(Ytrain)

Xtest=Xtest.reshape(10000,784)
Xtrain=Xtrain.reshape(60000,784)
Xtrain = Xtrain.astype("float32") / 255.0
Xtest = Xtest.astype("float32") / 255.0
```

```python
Ytest.shape
```

```python
from keras import Sequential
from keras.layers import Dense , Activation
model=Sequential()
model.add(Dense(128,activation='relu',input_shape=(784,)))
model.add(Dense(256,activation='relu'))
model.add(Dense(265,activation='relu'))
model.add(Dense(64,activation='relu'))
model.add(Dense(10,activation='softmax'))
# soft max bec >10class
```

```python
model.summary()
```

compile >> fit >>evsluate

```python
model.compile(loss="categorical_crossentropy",optimizer="adam",metrics=['accuracy']


)
```

```python
history=model.fit(Xtrain,Ytrain,validation_split=.1,epochs=100,batch_size=128)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')
plt.legend(['train', 'val'], loc='upper left')
plt.show()
```

```python
test_loss, test_acc = model.evaluate(Xtest,Ytest)
print('test_acc:', test_acc)
test_loss
```

```python

import numpy as np

# Get predictions from the trained model
y_pred_prob = model.predict(Xtest)

# Convert probabilities to digit labels
y_pred = np.argmax(y_pred_prob, axis=1)

# Convert actual one-hot labels to digit labels
y_true = np.argmax(Ytest, axis=1)

print("Actual labels:", y_true[:10])
print("Predicted labels:", y_pred[:10])

```

```python

import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_true, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=np.arange(10)
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Confusion Matrix - MNIST DNN")
plt.show()
```

```python

# Number of correct predictions for each digit
correct = np.diag(cm)

# Total actual samples for each digit
total = cm.sum(axis=1)

# Recall for each digit
recall = correct / total

for digit, score in enumerate(recall):
    print(f"Digit {digit}: Recall = {score:.2%}")
```
